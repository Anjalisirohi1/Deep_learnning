```python
import pandas as pd
```

```python
df = pd.read_csv('diabetes.csv')
```

```python
import numpy as np
```

```python
df.head()
```

```python
df.corr()['Outcome']
```

```python
X = df.iloc[:,:-1].values
y = df.iloc[:,-1].values
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X = scaler.fit_transform(X)
```

```python
X
```

```python
X.shape
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,_y_test = train_test_split(X,y,test_size=0.2,random_state=1)
```

```python
import tensorflow
from tensorflow import keras

```

```python
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()
model.add(Dense(32,activation='relu',input_dim=8))
model.add(Dense(1,activation='sigmoid'))

model.compile(optimizer='Adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model.fit(X_train,y_train,batch_size=32,epochs=100,validation_data=(X_test,_y_test))
```

```python
import keras_tuner as kt
```

```python
def build_model(hp):
    model =  Sequential()
    model.add(Dense(32,activation='relu',input_dim=8))
    model.add(Dense(1,activation='sigmoid'))
    model.compile(optimizer=hp.Choice('optimizer',values=['adam','sgd','rmsprop','adadelta']),loss='binary_crossentropy',metrics=['accuracy'])
    return model
```

```python
tuner = kt.RandomSearch(build_model,objective='val_accuracy',max_trials=5)
```

```python
tuner.search(X_train,y_train,epochs=5,validation_data=(X_test,_y_test))
```

```python
tuner.results_summary()
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
model = tuner.get_best_models(num_models=1)[0]
```

```python
model.summary()
```

```python
model.fit(X_train,y_train,batch_size=32,epochs=100,initial_epoch=6,validation_data=(X_test,_y_test))
```
