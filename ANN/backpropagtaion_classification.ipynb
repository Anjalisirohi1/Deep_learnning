```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame([[8,8,1],[7,9,1],[6,10,0],[5,12,0]],columns=['cgpa','profile_score','lpa'])
```

```python
def initialize_para(layer_dims):
  np.random.seed(3)
  parameters = {}
  L = len(layer_dims)

  for l in range(1,L):
    parameters['w'+str(l)] = np.ones((layer_dims[l-1],layer_dims[l]))*0.1
    parameters['b'+str(l)] = np.zeros((layer_dims[l],1))
  return parameters
```

```python
def sigmoid(Z):
  A = 1/(1+np.exp(-Z))
  return A
```

```python
def linear_forward(A_prev,w,b):
  Z = np.dot(w.T,A_prev)+b
  A = sigmoid(Z)

  return A
```

```python
def L_layer_forward(X,para):
  A = X
  L = len(para)//2
  for l in range(1,L+1):
    A_prev = A
    w1 = para['w'+str(l)]
    b1 = para['b'+str(l)]
    #print("A"+str(l-1)+": ",A_prev)
    #print("W"+str(l)+": ",w1)
    #print("b"+str(l)+": ",b1)
    #print("--"*20)

    A = linear_forward(A_prev,w1,b1)
    #print("A"+str(l)+": ",A)
    #print("**"*20)
  return A,A_prev

```

```python
def update_para(para,y,y_hat,A1,X):
  para['w2'][0][0] = para['w2'][0][0] + (0.001 * (y - y_hat)*A1[0][0])
  para['w2'][1][0] = para ['w2'][1][0] + (0.001 * (y - y_hat)*A1[1][0])
  para['b2'][0][0] = para['b2'][0][0] + (0.001 * (y - y_hat))

  para['w1'][0][0] = para['w1'][0][0] + (0.001 * (y - y_hat)*para['w2'][0][0]*A1[0][0]*(1-A1[0][0])*X[0][0])
  para['w1'][0][1] = para['w1'][0][1] + (0.001 * (y - y_hat)*para['w2'][0][0]*A1[0][0]*(1-A1[0][0])*X[1][0])
  para['b1'][0][0] = para['b1'][0][0] + (0.001 * (y - y_hat)*para['w2'][0][0]*A1[0][0]*(1-A1[0][0]))

  para['w1'][1][0] = para['w1'][1][0] + (0.001 * (y - y_hat)*para['w2'][1][0]*A1[1][0]*(1-A1[1][0])*X[0][0])
  para['w1'][1][1] = para['w1'][1][1] + (0.001 * (y - y_hat)*para['w2'][1][0]*A1[1][0]*(1-A1[1][0])*X[1][0])
  para['b1'][1][0] = para['b1'][1][0] + (0.001 * (y - y_hat)*para['w2'][1][0]*A1[1][0]*(1-A1[1][0]))

```

```python
X = df[['cgpa','profile_score']].values[0].reshape(2,1)
y = df[['lpa']].values[0][0]

para = initialize_para([2,2,1])

y_hat,A1 = L_layer_forward(X, para)
y_hat = y_hat[0][0]

update_para(para,y,y_hat,A1,X)

print('Loss for this student - ',-y*np.log(y_hat)-(1-y)*np.log(1-y_hat))

para

```

```python
#epoch implementation
para = initialize_para([2,2,1])
epochs = 50

for i in range(epochs):
  loss = []
  for j in range(df.shape[0]):
    X = df[['cgpa','profile_score']].values[j].reshape(2,1)
    y = df[['lpa']].values[j][0]

    y_hat,A1 = L_layer_forward(X, para)
    y_hat = y_hat[0][0]

    update_para(para,y,y_hat,A1,X)
    loss.append(-y*np.log(y_hat)-(1-y)*np.log(1-y_hat))
  print('Epoch - ',i+1,'Loss - ',np.array(loss).mean())

para
```
