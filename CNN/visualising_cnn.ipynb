```python
import tensorflow
from tensorflow import keras
from tensorflow.keras.applications.vgg16 import VGG16, preprocess_input,decode_predictions
```

```python
model = VGG16()
```

```python
model.summary()
```

```python
for i in range (len(model.layers)):
    if 'conv' not in model.layers[i].name:
        continue
    filters,biases = model.layers[i].get_weights()
    print("layer number",i,model.layers[i].name,filters.shape)
```

```python
filters,biases = model.layers[1].get_weights()
```

```python
f_min,f_max = filters.min(),filters.max()
filters = (filters - f_min) / (f_max - f_min)
```

```python
import matplotlib.pyplot as plt
import cv2
```

```python
n_filters = 6
ix = 1
fig = plt.figure(figsize=(15,10))
for i in range(n_filters):
    f = filters[:,:,:,i]
    for j in range(3):
        plt.subplot(n_filters,3,ix)
        plt.imshow(f[:,:,j],cmap = 'gray')
        ix+=1
plt.show()
```

```python
from keras.models import Model
from tensorflow.keras.preprocessing import image
import numpy as np
```

```python
model = Model(inputs = model.inputs, outputs = model.layers[1].output)
```

```python
img = image.load_img(r'C:\Users\anjal\Deep_learning\CNN\viratkohli.jpg',target_size = (224,224))
img = image.img_to_array(img)
img = np.expand_dims(img,axis = 0)
img = preprocess_input(img)
```

```python
features = model.predict(img)

fig = plt.figure(figsize=(20,15))
for i in range(1,features.shape[3]+1):
    plt.subplot(8,8,i)
    plt.imshow(features[0,:,:,i-1], cmap = 'gray')
plt.show()
```

```python
model2 = VGG16()
```

```python
layer_index = [2,5,9,13,17]
output = [model2.layers[i].output for i in layer_index]
model3 = Model(inputs =model2.inputs, outputs = output)
```

```python
feature_map = model3.predict(img)
for i,fmap in zip(layer_index,feature_map):
    fig = plt.figure(figsize=(20,15))
    fig.suptitle("Layer_{}".format(i),fontsize = 20)
    for i in range(1,features.shape[3]+1):
        plt.subplot(8,8,i)
        plt.imshow(fmap[0,:,:,i-1],cmap = 'gray')
plt.show()
```
