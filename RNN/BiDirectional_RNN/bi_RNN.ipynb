```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense,Input,LSTM,GRU,Bidirectional
from keras.datasets import imdb
```

```python
(x_train,y_train),(x_test,y_test) = imdb.load_data(num_words=10000)
```

```python
x_train = pad_sequences(x_train,maxlen=100,padding='post',truncating='post')
x_test = pad_sequences(x_test,maxlen=100,padding='post',truncating='post')
```

```python
model =Sequential()
model.add(Input(shape=(100,)))
model.add(Embedding(10000,32))
model.add(SimpleRNN(5))
model.add(Dense(1,activation='sigmoid'))
model.summary()
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model1 =Sequential()
model1.add(Input(shape=(100,)))
model1.add(Embedding(10000,32))
model1.add(Bidirectional(SimpleRNN(5)))
model1.add(Dense(1,activation='sigmoid'))
model1.summary()
model1.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model2 =Sequential()
model2.add(Input(shape=(100,)))
model2.add(Embedding(10000,32))
model2.add(Bidirectional(LSTM(5)))
model2.add(Dense(1,activation='sigmoid'))
model2.summary()
model2.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model3 =Sequential()
model3.add(Input(shape=(100,)))
model3.add(Embedding(10000,32))
model3.add(Bidirectional(GRU(5)))
model3.add(Dense(1,activation='sigmoid'))
model3.summary()
model3.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```
