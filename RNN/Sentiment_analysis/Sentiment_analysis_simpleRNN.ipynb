```python
import numpy as np 

docs = ['go india',
        'india india',
        'hip hip hurray',
        'jeetega bhai jeetega india jeetega',
        'bharat mata ki jai',
        'kohli kohli',
        'sachin sachin',
        'dhoni dhoni',
        'anjali ji ki jai',
        'inquilab zindabad']
```

```python
import tensorflow
from tensorflow import keras
from keras.datasets import imdb
from tensorflow.keras.preprocessing.text import Tokenizer
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
from keras import Sequential
```

```python
tokenizer =Tokenizer()
tokenizer.fit_on_texts(docs)
```

```python
len(tokenizer.word_index)
```

```python
sequences = tokenizer.texts_to_sequences(docs)
sequences
```

```python
from keras.utils import pad_sequences
```

```python
sequences = pad_sequences(sequences,padding='post')
sequences
```

```python
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten,Input
```

```python
model = Sequential()
vocab_size = len(tokenizer.word_index) + 1
model.add(Input(shape=(5,)))
model.add(Embedding(vocab_size,output_dim=2))
model.summary()
```

```python
model.compile('adam','accuracy')
```

```python
pred = model.predict(sequences)
print(pred)
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data(num_words=10000)
```

```python
X_train = pad_sequences(X_train,padding='post',maxlen=50)
X_test = pad_sequences(X_test,padding='post',maxlen=50)
```

```python
X_train.shape
```

```python
model1 = Sequential()
model1.add(Input(shape=(50,)))
model1.add(Embedding(10000,2))
model1.add(SimpleRNN(32,input_shape=(50,1),return_sequences=False))
model1.add(Dense(1,activation='sigmoid'))

model1.summary()
```

```python
model1.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
history = model1.fit(X_train,y_train,epochs=5,validation_data=(X_test,y_test))
```
